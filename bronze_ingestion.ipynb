```python
from pyspark.sql.functions import col, sum as spark_sum
print("Total employees:", employees_df.count())
```

```python
employees_df = spark.read.option("multiLine", "true").json(source_path)
print("Columns:", employees_df.columns)

duplicate_count = (
    employees_df.count() - employees_df.dropDuplicates().count()
)
print("Duplicate records:", duplicate_count)
```

```python
null_check = employees_df.select([
    spark_sum(col(c).isNull().cast("int")).alias(c)
    for c in employees_df.columns
])

display(null_check)
```
